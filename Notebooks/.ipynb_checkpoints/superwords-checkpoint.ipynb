# Superwords

Cet exercice va vous amener à écrire une fonction dont le but sera de transformer une phrase, selon certaines règles. Elle pourra changer quelque chose comme `Le Python, c'est super cool` en `L3 Py7h0n, C'357 5uP3R c00L`.

À première vue, il s’agit de remplacer certaines lettres par des chiffres dont la graphie est proche, mais regardez de plus près : certaines lettres sont mises en majuscule. Voici concrètement les règles de remplacement :


- indépendamment de la casse originale, la première lettre de la chaîne, qu’elle soit remplacée ou non, est mise en majuscule, et toutes les lettres suivantes alternent entre minuscules et majuscules. Par exemple, `Hello world` donnera, pour cette seule règle, `HeLlO wOrLd` (notez que l’espace ne compte pas comme une lettre). Cette règle est appliquée en premier, car elle s’applique sur les caractères d’origine.
- Les `i` et `I` sont remplacés par `1`
- Les `z` et `Z` sont remplacés par `2`
- Les `e` et `E` sont remplacés par `3`
- Les `a` et `A` sont remplacés par `4`
- Les `s` et `S` sont remplacés par `5`
- Les `g` et `G` sont remplacés par `6`
- Les `t` et `T` sont remplacés par `7`
- Les `b` et `B` sont remplacés par `8`
- les `o` et `O` sont remplacés par `0`

Votre rôle est donc d’écrire cette fonction, nommée `superwords`, qui prendra en argument une phrase, et retournera la version transformée de celle-ci.

```python
def conversionLettres(lettre):
    """
    Convertit une lettre en le chiffre si elle est dans la table.
    """
    correspondance = {
        "i": "1",
        "z": "2",
        "e": "3",
        "a": "4",
        "s": "5",
        "g": "6",
        "t": "7",
        "b": "8",
        "o": "0",
    }
    return correspondance.get(lettre.lower(), lettre) 


def superwords(texte):
    """
    Conversion des lettres en chiffres via conversionLettres() + alternance Maj/minuscule
    """
    resultat = []
    majuscule = True #Maj sur le premier caracter

    for caracter in texte: 
        if caracter.isalpha():
            caracter = caracter.upper() if majuscule else caracter.lower()
            majuscule = not majuscule
            caracter = conversionLettres(caracter)     

    return "".join(resultat)
```

Une fois votre fonction écrite, exécutez la cellule suivante pour savoir si elle fonctionne correctement.

```python
import sys
from itertools import count

tests: dict[str, str] = {
    'Hello world': 'H3Ll0 w0rLd',
    'uuu uu': 'UuU uU',
    'UUU UU': 'UuU uU',
    "Le Python, c'est cool !": "L3 Py7h0n, C'357 C00l !",
    'abcdefghijklmnopqrstuvwxyz': '48Cd3f6h1jKlMn0pQr57UvWxY2',
    'IIIIii': '111111',
    '"hey"': '"H3Y"',
    '"Aïe"': '"4ï3"',
    'Le développement, ça déchire': 'L3 DéV3L0Pp3m3n7, ç4 dÉcH1R3',
    "Σκεφτείτε να υποστηρίξετε άλλα αλφάβητα": "ΣκΕφΤεΊτΕ νΑ υΠοΣτΗρΊξΕτΕ άΛλΑ αΛφΆβΗτΑ",
}

passes = fails = errors = 0

for c, arg, expected in zip(count(1), tests.keys(), tests.values()):
    try:
        found = superwords(arg)
        print(f'Test #{c}...', flush=True)
        print(f'Argument :   "{arg}"', flush=True)
        print(f'Attendu :    "{expected}"', flush=True)
        assert found == expected, f'Obtenu :     "{found}"\n\N{CROSS MARK} le résultat est incorrect'
        print(f'\N{WHITE HEAVY CHECK MARK} Le résultat est correct', flush=True)
        passes += 1
    except AssertionError as e:
        print(e, file=sys.stderr, flush=True)
        fails += 1
    except Exception as e:
        print(f'Exception de type {type(e).__name__} lors de l\'appel de "{arg}" : {e}', file=sys.stderr, flush=True)
        errors += 1
    print(flush=True)

print(f'Résultats pour {len(tests)} tests : {passes} réussites, {fails} échecs, {errors} erreurs')
```
